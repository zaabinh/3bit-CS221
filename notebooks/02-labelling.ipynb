```python
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
df = pd.read_csv(".../Data/Restaurant_ABSA_aspect_corrections.csv")
len(df)
```

```python
df.columns
```

```python
main_df = pd.read_csv("../Data/Restaurant_ABSA_processed.csv")
len(main_df)
```

```python
for i in range(len(df)):
    source_index = df.loc[i, "source_row"] - 1
    for col in ["food", "price", "service", "ambiance", "miscellaneous"]:
        main_df.loc[source_index, col] = df.loc[i, col]
```

```python
main_df.to_csv("Data/Restaurant_ABSA_processed.csv", index=False)
```
