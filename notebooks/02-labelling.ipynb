```python
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
df = pd.read_csv("../data/temp/Restaurant_ABSA_aspect_corrections.csv")
len(df)
```

```python
main_df = pd.read_csv("../data/Restaurant_ABSA_processed.csv")
main_df.info()
```

```python
for i in range(len(df)):
    source_index = df.loc[i, "source_row"] 
    for col in ["food", "price", "service", "ambiance", "miscellaneous"]:
        main_df.loc[source_index, col] = df.loc[i, col]
```

```python
main_df.to_csv("../data/Restaurant_ABSA_processed.csv", index=False)
```
